```python
# Install yt-dlp if you haven't already
!pip install yt-dlp

# Install Node.js and npm to provide a JavaScript runtime for yt-dlp
!apt-get update
!apt-get install -y nodejs npm

# Ensure node is in a discoverable path, as yt-dlp might be picky
!ln -sf /usr/bin/nodejs /usr/local/bin/node

import yt_dlp
import os

def download_youtube_video(video_url, output_path='.', file_prefix=''):
    """
    Downloads a YouTube video using yt-dlp.

    Args:
        video_url (str): The URL of the YouTube video.
        output_path (str): The directory where the video will be saved.
        file_prefix (str): A prefix to add to the downloaded file name.
    """
    ydl_opts = {
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
        # Use video ID for outtmpl to ensure predictable and easily inferable filenames
        'outtmpl': os.path.join(output_path, f'{file_prefix}%(id)s.%(ext)s'),
        'noplaylist': True,  # Ensure only single video is downloaded even if part of a playlist
        'verbose': True,
        'progress': True,
        'js_runtimes': {'node': {'path': '/usr/local/bin/node'}}, # Explicitly provide the path to Node.js via symlink
    }

    print(f"\nAttempting to download: {video_url}")
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info_dict = ydl.extract_info(video_url, download=True)
            print(f"Successfully downloaded: {info_dict.get('title', 'Video')}")

            # Even if 'filepath' is missing, we can now reliably infer it
            # because outtmpl uses %(id)s.%(ext)s
            if 'filepath' not in info_dict:
                print(f"Warning: 'filepath' not found in info_dict for {video_url}. Inferring based on outtmpl.")
                video_id = info_dict.get('id')
                video_ext = info_dict.get('ext', 'mp4') # Default to mp4 if ext is missing
                if video_id and video_ext:
                    inferred_filepath = os.path.join(output_path, f"{file_prefix}{video_id}.{video_ext}")
                    info_dict['filepath'] = inferred_filepath
                else:
                    print(f"Error: Could not infer filepath for {video_url} as ID or extension was missing.")
                    return None # Return None if filepath cannot be determined

            return info_dict
    except Exception as e:
        print(f"Error downloading {video_url}: {e}")
        return None

# --- Example Usage --- #
# Replace with the actual video URL you want to download
# For downloading from a channel, you would typically get individual video URLs
# For 'https://www.youtube.com/@TigerVlogs. (www.youtube.com)', you would need to iterate through its videos.
# You would typically get these links from your Google Sheet.

# Example video URL (replace with actual video link)
# For demonstration, I'm using a placeholder. You would fetch these from your Google Sheet.
example_video_url = 'https://www.youtube.com/watch?v=dQw4w9WgXcQ' # A sample public YouTube video

# Specify an output directory (optional, defaults to current directory)
output_directory = 'downloaded_videos'
os.makedirs(output_directory, exist_ok=True)

# Download the video
download_info = download_youtube_video(example_video_url, output_path=output_directory, file_prefix='TigerVlogs_')

# You can access downloaded file path, title, etc. from download_info
# if download_info:
#     print(f"Download complete. File saved to: {download_info.get('filepath')}")
#     print(f"Video Title: {download_info.get('title')}")

# --- Next Steps / Considerations for your workflow --- #
# 1. Google Sheets Integration: Use libraries like `gspread` to read video URLs from your Google Sheet.
# 2. Iterate and Download: Loop through the URLs from Google Sheet and call `download_youtube_video` for each.
# 3. Video Analysis/Metadata: `yt-dlp` can also extract metadata without downloading the video. Further analysis would require other libraries (e.g., OpenCV, moviepy, or even AI models).
# 4. Content Modification: Python string operations for title/description/keyword updates. AI for new thumbnail ideas.
# 5. YouTube Upload: Requires the Google API Client Library for Python and YouTube Data API v3.
# 6. make.com Integration: This would involve make.com's own modules or webhooks to trigger actions or update data after processing in Colab).
```

### Note on `cookies.txt` for Private Videos

To download private or age-restricted YouTube videos, `yt-dlp` often requires authentication cookies. You'll need to export your browser's YouTube cookies into a file named `cookies.txt` and upload it to your Colab environment (in the same directory as this notebook).

**How to get `cookies.txt`:**
1.  Install a browser extension like "Get cookies.txt" (for Chrome, Firefox, etc.).
2.  Go to YouTube (make sure you are logged in to the account that has access to the private videos).
3.  Use the extension to download the `cookies.txt` file.
4.  Upload this `cookies.txt` file to your Colab workspace using the file browser on the left.

```python
import yt_dlp
import os

def download_youtube_video(video_url, output_path='.', file_prefix='', cookiefile='/content/cookies.txt'):
    """
    Downloads a YouTube video using yt-dlp.

    Args:
        video_url (str): The URL of the YouTube video.
        output_path (str): The directory where the video will be saved.
        file_prefix (str): A prefix to add to the downloaded file name.
        cookiefile (str): Path to a Netscape-style cookies file for authentication.
    """
    ydl_opts = {
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
        'outtmpl': os.path.join(output_path, f'{file_prefix}%(id)s.%(ext)s'),
        'noplaylist': True,
        'verbose': True,
        'progress': True,
        'js_runtimes': {'node': {'path': '/usr/local/bin/node'}},
    }

    # Add cookiefile option if the file exists
    if os.path.exists(cookiefile):
        ydl_opts['cookiefile'] = cookiefile
        print(f"Using cookiefile: {cookiefile}")
    else:
        print(f"Warning: Cookie file '{cookiefile}' not found. Downloads of private or age-restricted videos may fail.")

    print(f"\nAttempting to download: {video_url}")
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info_dict = ydl.extract_info(video_url, download=True)
            print(f"Successfully downloaded: {info_dict.get('title', 'Video')}")

            if 'filepath' not in info_dict:
                print(f"Warning: 'filepath' not found in info_dict for {video_url}. Inferring based on outtmpl.")
                video_id = info_dict.get('id')
                video_ext = info_dict.get('ext', 'mp4')
                if video_id and video_ext:
                    inferred_filepath = os.path.join(output_path, f"{file_prefix}{video_id}.{video_ext}")
                    info_dict['filepath'] = inferred_filepath
                else:
                    print(f"Error: Could not infer filepath for {video_url} as ID or extension was missing.")
                    return None

            return info_dict
    except Exception as e:
        print(f"Error downloading {video_url}: {e}")
        return None

# --- Example Usage (will be handled by subsequent cells) --- #
```

### Step 4: Filter and Sort Videos for Download

Before downloading, we need to filter the `df_videos` DataFrame to get videos that haven't been uploaded (`IsUploaded` is empty or false), and then sort them by their publish date to ensure we process oldest first. We'll extract the publish date during the `yt-dlp` info extraction step. For now, we'll assume `IsUploaded` is either empty or 'false'.

```python
# Filter for videos that have not yet been uploaded
# Assuming 'IsUploaded' column will be empty or 'FALSE' for not uploaded videos
df_pending_uploads = df_videos[df_videos['IsUploaded'].fillna('') == ''].copy()

# To sort by publish date, we'll need to get the publish date for each video.
# We can use yt-dlp's info extraction feature without downloading the video.
# Let's get info for the first few pending videos to extract publish date.

video_urls_to_process = []
videos_with_metadata = []

for index, row in df_pending_uploads.head(5).iterrows(): # Process first 5 pending to get some publish dates
    video_url = row['Video URL']
    print(f"Fetching info for {video_url}")
    try:
        with yt_dlp.YoutubeDL({'noplaylist': True, 'quiet': True}) as ydl:
            info_dict = ydl.extract_info(video_url, download=False)
            publish_date_str = info_dict.get('upload_date') # Format YYYYMMDD
            if publish_date_str:
                # Convert YYYYMMDD to datetime object
                publish_date = pd.to_datetime(publish_date_str, format='%Y%m%d')
                row['Publish_Date'] = publish_date
                videos_with_metadata.append(row)
                video_urls_to_process.append(video_url)
            else:
                print(f"Could not get publish date for {video_url}")
    except Exception as e:
        print(f"Error fetching info for {video_url}: {e}")

if videos_with_metadata:
    df_processed_with_dates = pd.DataFrame(videos_with_metadata)
    df_processed_with_dates = df_processed_with_dates.sort_values(by='Publish_Date', ascending=True)

    # Select the top 3 oldest videos to download
    videos_to_download_temp = df_processed_with_dates.head(3).copy() # Use .copy() to avoid SettingWithCopyWarning

    # Now, let's use the download_youtube_video function for these 3 videos and store file paths
    downloaded_video_paths = []
    for index, video_row in videos_to_download_temp.iterrows():
        print(f"\nDownloading video: {video_row['Video Title']}")
        download_info = download_youtube_video(
            video_row['Video URL'],
            output_path=output_directory,
            file_prefix='TigerVlogs_'
        )
        if download_info and 'filepath' in download_info:
            downloaded_video_paths.append({
                'Video ID': video_row['Video ID'],
                'Actual File Path': download_info['filepath'] # Store the exact file path
            })
        else:
            print(f"Warning: Failed to download or get filepath for {video_row['Video URL']}")

    if downloaded_video_paths:
        df_actual_paths = pd.DataFrame(downloaded_video_paths)
        # Merge the actual file paths back into videos_to_download_temp
        videos_for_download = pd.merge(videos_to_download_temp, df_actual_paths, on='Video ID', how='left')
        print("\nTop 3 oldest videos successfully downloaded with their actual file paths:")
        display(videos_for_download[['Video Title', 'Video URL', 'Publish_Date', 'Actual File Path']])
    else:
        print("\nNo videos were successfully downloaded with file paths.")
        videos_for_download = pd.DataFrame() # Ensure it's empty if no downloads succeeded

else:
    print("No pending videos with valid publish dates found to process.")
```

```python
import os
import pickle
import google.auth.transport.requests
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
from googleapiclient.http import MediaFileUpload

# The CLIENT_SECRETS_FILE contains your OAuth 2.0 Client ID and Client Secret.
CLIENT_SECRETS_FILE = 'youtube_client_secret.json'

# This scope allows for full upload access to the authenticated user's YouTube channel.
SCOPES = ['https://www.googleapis.com/auth/youtube.upload']
API_SERVICE_NAME = 'youtube'
API_VERSION = 'v3'

def get_authenticated_service():
    credentials = None

    # 1. Check if token.pickle exists and verify required scopes
    if os.path.exists('token.pickle'):
        try:
            with open('token.pickle', 'rb') as token:
                credentials = pickle.load(token)
            token_scopes = getattr(credentials, 'scopes', []) or []
            if token_scopes and not any('youtube' in s for s in token_scopes):
                print("Existing token.pickle has insufficient scopes. Clearing stale token...")
                credentials = None
                os.remove('token.pickle')
        except Exception as e:
            print(f"Could not load existing token.pickle: {e}. Re-authenticating...")
            credentials = None

    # 2. Refresh credentials if expired
    if credentials and credentials.expired and credentials.refresh_token:
        try:
            credentials.refresh(google.auth.transport.requests.Request())
        except Exception as e:
            print(f"Failed to refresh token: {e}. Re-authenticating...")
            credentials = None

    # 3. If credentials are not available/valid, initiate OAuth flow
    if not credentials or not credentials.valid:
        if not os.path.exists(CLIENT_SECRETS_FILE):
            raise FileNotFoundError(
                f"'{CLIENT_SECRETS_FILE}' not found! Please upload your client secret JSON or token.pickle to the Colab files panel."
            )

        flow = InstalledAppFlow.from_client_secrets_file(
            CLIENT_SECRETS_FILE,
            scopes=SCOPES,
            redirect_uri='http://localhost'
        )

        try:
            # Works in local desktop Python environments
            credentials = flow.run_local_server(port=0)
        except Exception:
            # Fallback for Google Colab and remote headless environments
            auth_url, _ = flow.authorization_url(prompt='consent', access_type='offline')
            print("\n" + "="*70)
            print("AUTH REQUIRED: Please open this URL in your browser to authorize:")
            print(auth_url)
            print("="*70)
            print("1. Sign in with your YouTube Google Account and click Allow.")
            print("2. Your browser will redirect to localhost (which may show 'Site can't be reached').")
            print("3. Copy the ENTIRE URL from your browser's address bar and paste it below:")
            auth_response = input("Paste redirected URL (or code) here: ").strip()

            if auth_response.startswith('http'):
                flow.fetch_token(authorization_response=auth_response)
            else:
                flow.fetch_token(code=auth_response)

            credentials = flow.credentials

        # Save valid credentials for future runs
        with open('token.pickle', 'wb') as token:
            pickle.dump(credentials, token)
        print("Credentials saved to 'token.pickle'.")

    return build(API_SERVICE_NAME, API_VERSION, credentials=credentials)

# Authenticate and build the YouTube service object
youtube = get_authenticated_service()
print("YouTube Data API authenticated successfully!")
```
