```python
#@title 🔐 Step 2: Verify Channel Credentials & YouTube Token
import os, json

token_candidates = [
    'channels/MindNestTales/token.json',
    'channels/TigerVlogs-2026/token.json',
    'token.json'
]
found_token = None
for t in token_candidates:
    if os.path.exists(t):
        found_token = t
        break

if found_token:
    print(f"✅ YouTube token detected at: {found_token}")
    # Copy to MindNestTales if not present
    dest = 'channels/MindNestTales/token.json'
    if found_token != dest:
        with open(found_token, 'r') as src, open(dest, 'w') as dst:
            dst.write(src.read())
        print(f"Copied token to {dest}")
else:
    print("⚠️ No token.json found yet.")
    print("If running in Colab, you can upload your token.json to channels/MindNestTales/token.json via the Files tab on the left.")
```

```python
#@title 🎬 Step 3: Interactive Story & Quality Selector { run: "auto" }
#@markdown Select story and target release settings:
story_number = 1 #@param {type:"integer"}
target_language = "all" #@param ["all", "hi", "en", "es", "pt", "fr", "de", "ar", "ja", "ko", "vi", "fil"]
quality = "4k" #@param ["8k", "4k", "1080p"]
auto_advance_next_day = True #@param {type:"boolean"}

print(f"Selected: Story #{story_number} | Language: {target_language.upper()} | Quality: {quality.upper()} Shorts")
```

```python
#@title 🚀 Step 4: Execute 8K/4K Shorts Production & Multi-Channel Upload
import subprocess, sys

cmd = [
    sys.executable,
    "channels/MindNestTales/scheduler.py",
    str(story_number),
    target_language,
    quality
]

print(f"Executing: {' '.join(cmd)}\n")
subprocess.run(cmd, check=True)
```
