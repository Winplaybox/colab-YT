```python
#@title 🔐 Step 2: Verify Channel Credentials
import os, json

token_candidates = [
    'channels/AllTrainReels/token.json',
    'token.json',
    'channels/IndianHorrorTapes/token.json'
]
token_found = any(os.path.exists(p) for p in token_candidates)
if token_found:
    print("✅ Active YouTube OAuth Token detected!")
else:
    print("⚠️ No token.json found yet. You can render videos in simulation/preview mode or upload token.json.")
```

```python
#@title 🎬 Step 3: Select Story, Visual Theme & Quality { run: "auto" }
#@markdown Choose any episode number from 1 to 30:
episode_number = 1 #@param {type:"integer"}
visual_theme = "Auto (Theme-Matching)" #@param ["Auto (Theme-Matching)", "SPEEDOMETER_HUD", "SPLIT_VS_ARENA", "VINTAGE_STEAM_GAUGE", "CABIN_POV_RADAR", "ARCHITECTURAL_BLUEPRINT"]
quality = "1080p" #@param ["1080p", "4k", "8k"]
publish_mode = "Schedule Daily 6:30 PM IST" #@param ["Schedule Daily 6:30 PM IST", "Publish Immediately", "Render Only (Preview)"]

sys.path.insert(0, 'channels/AllTrainReels')
from story_generator import get_or_create_story
story = get_or_create_story(episode_number)

if visual_theme != "Auto (Theme-Matching)":
    story["visual_theme"] = visual_theme

print(f"🔥 Selected Episode #{story['episode_number']}: {story['title']}")
print(f"🚂 Format: {story.get('format')}")
print(f"🎨 Visual Theme: {story.get('visual_theme')}")
print(f"🎯 Target Quality: {quality.upper()} Portrait (9:16)")
print(f"⏱️ Duration: ~{story.get('target_duration_seconds', 25)}s")
print(f"🚀 Mode: {publish_mode}")
```

```python
#@title ⚡ Step 4: Render High-Speed 9:16 Short
import os, glob
from video_producer import produce_train_video
from IPython.display import Video, display

story_files = glob.glob(f"channels/AllTrainReels/stories/story_{episode_number:02d}_*.json")
if not story_files:
    raise FileNotFoundError(f"Story file for Episode #{episode_number} not found. Run Step 3 first.")

story_path = story_files[0]
print(f"Rendering from: {story_path}")
video_path = produce_train_video(story_path, quality=quality)

print(f"\n🎉 Render Complete: {video_path}")
if os.path.exists(video_path):
    display(Video(video_path, embed=True, width=360))
```

```python
#@title 🚀 Step 5: Upload & Schedule to YouTube Shorts
from scheduler import upload_short
import json

with open(story_path, 'r', encoding='utf-8') as f:
    story_data = json.load(f)

mode_arg = "schedule" if "Schedule" in publish_mode else ("public" if "Immediately" in publish_mode else "render_only")

if mode_arg == "render_only":
    print("ℹ️ Preview mode selected. Skipping YouTube upload.")
else:
    video_id = upload_short(video_path, story_data, publish_mode=mode_arg)
    if video_id:
        print(f"✅ Successfully processed YouTube Short: https://www.youtube.com/shorts/{video_id}")
```
